```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report, confusion_matrix

from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.naive_bayes import GaussianNB
import matplotlib.pyplot as plt
import seaborn as sns

# =========================
# 1. LOAD + BASIC CLEANING
# =========================
file_path = "../../data/ckd/kidney_disease.csv"
DATASET_NAME = "kidney_disease"
df = pd.read_csv(file_path)

df = df.replace(r'^\s+|\s+$', '', regex=True)
df = df.replace('?', np.nan)
df = df.replace('\t?', np.nan)
df = df.replace({'\tyes': 'yes', '\tno': 'no', ' yes': 'yes', ' no': 'no'})

# find target
possible_targets = ['classification', 'class', 'target', 'label']
target_col = next((c for c in possible_targets if c in df.columns), None)
if target_col is None:
    raise ValueError("Target column not found.")

df[target_col] = df[target_col].astype(str).str.strip().str.lower()
df[target_col] = df[target_col].replace({
    'ckd\t': 'ckd',
    'notckd': 'notckd',
    'not ckd': 'notckd'
})

df = df[df[target_col].isin(['ckd', 'notckd'])].copy()

X = df.drop(columns=[target_col]).copy()
y = df[target_col].copy()

X = X.drop(columns=['id'])

#  numeric conversion
for col in X.columns:
    X[col] = pd.to_numeric(X[col], errors='ignore')

numeric_cols = X.select_dtypes(include=['number']).columns.tolist()
categorical_cols = [c for c in X.columns if c not in numeric_cols]

# target encode
y = y.map({'ckd': 1, 'notckd': 0})

print("Shape:", X.shape)
print("Target counts:\n", y.value_counts())
print("Numeric:", numeric_cols)
print("Categorical:", categorical_cols)

# =========================
# 2. SPLIT FIRST (70:30)
# =========================
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.30,
    random_state=42,
    stratify=y
)

# =========================
# 3. IMPUTATION USING TRAIN ONLY
# =========================
rng = np.random.default_rng(42)

def random_sample_impute_train_test(train_s, test_s, rng_obj):
    train_s = train_s.copy()
    test_s = test_s.copy()

    observed = train_s.dropna().values
    if len(observed) == 0:
        return train_s, test_s

    # impute train from train observed values
    train_missing = train_s.isna()
    if train_missing.sum() > 0:
        sampled_train = rng_obj.choice(observed, size=train_missing.sum(), replace=True)
        train_s.loc[train_missing] = sampled_train

    # recompute observed after train fill
    observed = train_s.dropna().values

    # impute test from train distribution
    test_missing = test_s.isna()
    if test_missing.sum() > 0:
        sampled_test = rng_obj.choice(observed, size=test_missing.sum(), replace=True)
        test_s.loc[test_missing] = sampled_test

    return train_s, test_s

# numeric -> random sample, fallback mean
for col in numeric_cols:
    try:
        X_train[col], X_test[col] = random_sample_impute_train_test(X_train[col], X_test[col], rng)
    except Exception:
        mean_val = X_train[col].mean()
        X_train[col] = X_train[col].fillna(mean_val)
        X_test[col] = X_test[col].fillna(mean_val)

# categorical -> random sample, fallback mode
for col in categorical_cols:
    try:
        X_train[col], X_test[col] = random_sample_impute_train_test(X_train[col], X_test[col], rng)
    except Exception:
        mode_val = X_train[col].mode(dropna=True)
        if len(mode_val) > 0:
            fill_val = mode_val.iloc[0]
            X_train[col] = X_train[col].fillna(fill_val)
            X_test[col] = X_test[col].fillna(fill_val)

# final fallback
for col in numeric_cols:
    mean_val = X_train[col].mean()
    X_train[col] = X_train[col].fillna(mean_val)
    X_test[col] = X_test[col].fillna(mean_val)

for col in categorical_cols:
    mode_val = X_train[col].mode(dropna=True)
    if len(mode_val) > 0:
        fill_val = mode_val.iloc[0]
        X_train[col] = X_train[col].fillna(fill_val)
        X_test[col] = X_test[col].fillna(fill_val)

# =========================
# 4. CATEGORICAL ENCODING USING TRAIN ONLY
# =========================

from sklearn.preprocessing import OrdinalEncoder

encoder = OrdinalEncoder(
    handle_unknown="use_encoded_value",
    unknown_value=-1
)

X_train[categorical_cols] = encoder.fit_transform(
    X_train[categorical_cols].astype(str)
)

X_test[categorical_cols] = encoder.transform(
    X_test[categorical_cols].astype(str)
)

# =========================
# 5. MODELS
# =========================
models = {
    "KNN": KNeighborsClassifier(n_neighbors=5),
    "DT": DecisionTreeClassifier(random_state=42),
    "RF": RandomForestClassifier(n_estimators=200, random_state=42),
    "NB": GaussianNB(),
    "GB": GradientBoostingClassifier(random_state=42)
}

results = []

for name, model in models.items():
    # no scaling for strict replication-style run
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)

    acc = accuracy_score(y_test, y_pred)
    prec = precision_score(y_test, y_pred, zero_division=0)
    rec = recall_score(y_test, y_pred, zero_division=0)
    f1 = f1_score(y_test, y_pred, zero_division=0)

    print(f"\n=== {name} ===")
    print("Accuracy :", round(acc, 4))
    print("Precision:", round(prec, 4))
    print("Recall   :", round(rec, 4))
    print("F1 Score :", round(f1, 4))
    print(confusion_matrix(y_test, y_pred))

    results.append({
        "CLASSIFIERS": name,
        "ACCURACY (%)": round(acc * 100, 2),
        "PRECISION": round(prec, 4),
        "RECALL": round(rec, 4),
        "F1-SCORE": round(f1, 4)
    })

results_df = pd.DataFrame(results).sort_values("ACCURACY (%)", ascending=False)

print("\n=== TABLE 1 STYLE ===")
print(results_df[["CLASSIFIERS", "ACCURACY (%)"]])

print("\n=== FULL METRICS ===")
print(results_df)
```

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.naive_bayes import GaussianNB

# =========================
# 1. LOAD + BASIC CLEANING
# =========================
file_path = "../../data/ckd/kidney_disease.csv"
df = pd.read_csv(file_path)

df = df.replace(r'^\s+|\s+$', '', regex=True)
df = df.replace('?', np.nan)
df = df.replace('\t?', np.nan)
df = df.replace({'\tyes': 'yes', '\tno': 'no', ' yes': 'yes', ' no': 'no'})

possible_targets = ['classification', 'class', 'target', 'label']
target_col = next((c for c in possible_targets if c in df.columns), None)
if target_col is None:
    raise ValueError("Target column not found.")

df[target_col] = df[target_col].astype(str).str.strip().str.lower()
df[target_col] = df[target_col].replace({
    'ckd\t': 'ckd',
    'notckd': 'notckd',
    'not ckd': 'notckd'
})

df = df[df[target_col].isin(['ckd', 'notckd'])].copy()

X = df.drop(columns=[target_col]).copy()
y = df[target_col].map({'ckd': 1, 'notckd': 0})

# drop id
if 'id' in X.columns:
    X = X.drop(columns=['id'])

# numeric conversion where possible
for col in X.columns:
    X[col] = pd.to_numeric(X[col], errors='ignore')

numeric_cols = X.select_dtypes(include=['number']).columns.tolist()
categorical_cols = [c for c in X.columns if c not in numeric_cols]

print("Shape:", X.shape)
print("Target counts:\n", y.value_counts())
print("Numeric:", numeric_cols)
print("Categorical:", categorical_cols)

# =========================
# 2. LEAKAGE PREPROCESSING
# preprocess BEFORE split
# =========================
rng = np.random.default_rng(42)

def random_sample_impute(series: pd.Series, rng_obj) -> pd.Series:
    s = series.copy()
    missing_mask = s.isna()
    observed = s.dropna().values

    if missing_mask.sum() == 0 or len(observed) == 0:
        return s

    sampled = rng_obj.choice(observed, size=missing_mask.sum(), replace=True)
    s.loc[missing_mask] = sampled
    return s

# random-sample imputation on FULL DATASET
for col in numeric_cols:
    try:
        X[col] = random_sample_impute(X[col], rng)
    except Exception:
        X[col] = X[col].fillna(X[col].mean())

for col in categorical_cols:
    try:
        X[col] = random_sample_impute(X[col], rng)
    except Exception:
        mode_val = X[col].mode(dropna=True)
        if len(mode_val) > 0:
            X[col] = X[col].fillna(mode_val.iloc[0])

# final fallback
for col in numeric_cols:
    X[col] = X[col].fillna(X[col].mean())

for col in categorical_cols:
    mode_val = X[col].mode(dropna=True)
    if len(mode_val) > 0:
        X[col] = X[col].fillna(mode_val.iloc[0])

# label encoding on FULL DATASET
encoders = {}
for col in categorical_cols:
    le = LabelEncoder()
    X[col] = le.fit_transform(X[col].astype(str))
    encoders[col] = le

# =========================
# 3. SPLIT AFTER PREPROCESSING
# =========================
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.30,
    random_state=42,
    stratify=y
)

# =========================
# 4. MODELS
# =========================
models = {
    "KNN": KNeighborsClassifier(n_neighbors=5),
    "DT": DecisionTreeClassifier(random_state=42),
    "RF": RandomForestClassifier(n_estimators=200, random_state=42),
    "NB": GaussianNB(),
    "GB": GradientBoostingClassifier(random_state=42)
}

results_leakage = []

for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)

    acc = accuracy_score(y_test, y_pred)
    prec = precision_score(y_test, y_pred, zero_division=0)
    rec = recall_score(y_test, y_pred, zero_division=0)
    f1 = f1_score(y_test, y_pred, zero_division=0)

    print(f"\n=== {name} (LEAKAGE VERSION) ===")
    print("Accuracy :", round(acc, 4))
    print("Precision:", round(prec, 4))
    print("Recall   :", round(rec, 4))
    print("F1 Score :", round(f1, 4))

    results_leakage.append({
        "CLASSIFIERS": name,
        "ACCURACY (%)": round(acc * 100, 2),
        "PRECISION": round(prec, 4),
        "RECALL": round(rec, 4),
        "F1-SCORE": round(f1, 4)
    })

results_leakage_df = pd.DataFrame(results_leakage).sort_values("ACCURACY (%)", ascending=False)

print("\n=== LEAKAGE TABLE ===")
print(results_leakage_df)
```
